# Pandas Basics


### 1. Import Pandas


```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

### 2. Series & DataFrame Creation


```python
# Series
s = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
print("Series:\n", s)

# DataFrame from dictionary
data = {
    'Name': ['Ali', 'Sara', 'Bilal', 'Zainab'],
    'Age': [25, 28, 22, 30],
    'City': ['Karachi', 'Lahore', 'Islamabad', 'Karachi'],
    'Salary': [65000, 72000, 50000, 85000]
}
df = pd.DataFrame(data)
print("\nDataFrame:\n", df)
```

### 3. Inspecting Data


```python
print("Head:\n", df.head(2))
print("\nTail:\n", df.tail(2))
print("\nShape:", df.shape)
print("Columns:", list(df.columns))
print("Data types:\n", df.dtypes)
print("\nSummary statistics:\n", df.describe())
```

### 4. Selecting Columns & Rows (loc & iloc)


```python
# Single column
print("Single column (Salary):\n", df['Salary'])

# Multiple columns
print("\nMultiple columns:\n", df[['Name', 'Salary']])

# Row selection by index number (iloc)
print("\nFirst row by iloc:\n", df.iloc[0])

# Specific cell (row 1, column 'Name') with loc
print("\nValue at row 1, Name with loc:", df.loc[1, 'Name'])

# Slicing rows 1 to 2
print("\nRows 1 to 2:\n", df.iloc[1:3])
```

### 5. Filtering Data


```python
# Single condition
print("Age > 24:\n", df[df['Age'] > 24])

# Multiple conditions
print("\nSalary > 60000 and City is Karachi:\n", df[(df['Salary'] > 60000) & (df['City'] == 'Karachi')])

# isin filter
print("\nCity in ['Lahore', 'Islamabad']:\n", df[df['City'].isin(['Lahore', 'Islamabad'])])
```

### 6. Adding & Dropping Columns


```python
# Add new column
df['Tax'] = df['Salary'] * 0.10
df['Net_Salary'] = df['Salary'] - df['Tax']
print("After adding Tax and Net_Salary:\n", df)

# Drop a column
df_dropped = df.drop(columns=['Tax'])
print("\nAfter dropping Tax:\n", df_dropped)
```

### 7. Handling Missing Values


```python
sample = pd.DataFrame({
    'Item': ['A', 'B', 'C', 'D'],
    'Price': [10, np.nan, 30, np.nan],
    'Stock': [100, 200, np.nan, 400]
})
print("Sample with NaNs:\n", sample)
print("\nMissing count per column:\n", sample.isnull().sum())

# Fill missing values
filled = sample.copy()
filled['Price'] = filled['Price'].fillna(filled['Price'].median())
filled['Stock'] = filled['Stock'].fillna(0)
print("\nAfter fillna:\n", filled)
```

### 8. Sorting & Value Counts


```python
# Sort values
print("Sorted by Salary descending:\n", df.sort_values(by='Salary', ascending=False))

# Value counts
print("\nCity counts:\n", df['City'].value_counts())
print("\nUnique cities:", df['City'].unique())
print("Number of unique cities:", df['City'].nunique())
```

### 9. GroupBy & Aggregations


```python
dept_data = pd.DataFrame({
    'Dept': ['IT', 'HR', 'IT', 'Finance', 'HR', 'IT'],
    'Salary': [70000, 50000, 85000, 65000, 55000, 90000]
})

print("Average salary by department:\n", dept_data.groupby('Dept')['Salary'].mean())
print("\nMultiple aggregations:\n", dept_data.groupby('Dept')['Salary'].agg(['count', 'mean', 'min', 'max']))
```

### 10. String Methods


```python
names = pd.Series(['  babar azam ', 'RIZWAN', 'fakhar zaman'])

print("Cleaned lowercase:", names.str.strip().str.title())
print("Contains 'zaman':", names.str.contains('zaman'))
```
