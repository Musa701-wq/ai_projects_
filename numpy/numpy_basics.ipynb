# NumPy Basics


### 1. Import NumPy


```python
import numpy as np

print("NumPy version:", np.__version__)
```

### 2. Array Creation & Attributes


```python
a = np.array([10, 20, 30, 40])
print("1D array:", a)

b = np.array([[1, 2, 3], [4, 5, 6]], dtype=float)
print("\n2D array:\n", b)

print("\nShape:", b.shape)
print("Dimensions:", b.ndim)
print("Data type:", b.dtype)
print("Total elements:", b.size)
```

### 3. Built-in Functions


```python
print("Zeros:\n", np.zeros((2, 3)))
print("\nOnes:\n", np.ones((2, 2), dtype=int))
print("\nFull with 7:\n", np.full((2, 3), 7))
print("\nIdentity matrix:\n", np.eye(3))
print("\narange:", np.arange(1, 10, 2))
print("linspace:", np.linspace(0, 1, 5))
```

### 4. Random Numbers


```python
np.random.seed(42)

print("Random integers:\n", np.random.randint(1, 50, size=(2, 3)))
print("\nRandom floats:", np.random.rand(4))
```

### 5. Indexing & Slicing


```python
arr = np.array([10, 20, 30, 40, 50, 60, 70])

print("arr[0]:", arr[0])
print("arr[-1]:", arr[-1])
print("arr[2:6]:", arr[2:6])
print("arr[::-1]:", arr[::-1])

mat = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("\nMatrix:\n", mat)
print("Row 0:", mat[0, :])
print("Col 1:", mat[:, 1])
print("Sub-matrix:\n", mat[:2, :2])
```

### 6. Reshape, Flatten & Transpose


```python
x = np.arange(1, 7)
print("Original:", x)

r = x.reshape(2, 3)
print("\nReshaped (2x3):\n", r)
print("\nTranspose:\n", r.T)
print("\nFlattened:", r.flatten())
```

### 7. Joining & Stacking


```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print("vstack:\n", np.vstack([arr1, arr2]))
print("\nhstack:", np.hstack([arr1, arr2]))

m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])

print("\nConcatenate axis 0:\n", np.concatenate([m1, m2], axis=0))
print("\nConcatenate axis 1:\n", np.concatenate([m1, m2], axis=1))
```

### 8. Math & Universal Functions


```python
n = np.array([4, 9, 16, 25])

print("Array + 10:", n + 10)
print("Array * 2:", n * 2)
print("Square root:", np.sqrt(n))
print("Log:", np.log(n).round(2))
print("Absolute values:", np.abs(np.array([-5, 10, -20])))
```

### 9. Stats Functions & Axis


```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Data:\n", data)
print("Total sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data).round(2))
print("Min:", np.min(data), "| Max:", np.max(data))

print("\nColumn sum (axis=0):", np.sum(data, axis=0))
print("Row sum (axis=1):", np.sum(data, axis=1))
```

### 10. Conditions & Filtering


```python
marks = np.array([35, 72, 85, 45, 90, 60])
print("Marks:", marks)

print("Marks >= 60:", marks >= 60)
print("Passed marks:", marks[marks >= 60])
print("Between 50 and 80:", marks[(marks >= 50) & (marks <= 80)])

result = np.where(marks >= 50, "Pass", "Fail")
print("Result:", result)
```

### 11. Searching & Sorting


```python
vals = np.array([40, 10, 30, 20, 10, 30, 50])
print("Values:", vals)

print("argmax:", np.argmax(vals))
print("argmin:", np.argmin(vals))
print("Sorted:", np.sort(vals))
print("argsort:", np.argsort(vals))
print("Unique:", np.unique(vals))
```

### 12. Matrix Multiplication


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("np.dot(A, B):\n", np.dot(A, B))
print("\nA @ B:\n", A @ B)
```
