# Matrix Operations in NumPy


### 1. Matrix Creation & Special Matrices


```python
import numpy as np

# Standard 2D Matrix (2 rows x 3 cols)
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("Matrix A (2x3):\n", A)
print("Shape:", A.shape)

# Special Matrices
identity = np.eye(3, dtype=int)         # 3x3 Identity matrix
zeros = np.zeros((2, 2), dtype=int)     # 2x2 Zero matrix
diag_matrix = np.diag([10, 20, 30])    # Diagonal matrix

print("\nIdentity Matrix (3x3):\n", identity)
print("\nZero Matrix (2x2):\n", zeros)
print("\nDiagonal Matrix:\n", diag_matrix)
```

### 2. Matrix Addition, Subtraction & Scalar Multiplication


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

# Addition & Subtraction (must be same shape)
print("\nA + B:\n", A + B)
print("\nA - B:\n", A - B)

# Scalar Multiplication
print("\n3 * A:\n", 3 * A)
```

### 3. Element-wise Multiplication vs Matrix Multiplication (@)


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Element-wise multiplication (Hadamard product: a_ij * b_ij)
elem_prod = A * B
print("Element-wise (A * B):\n", elem_prod)

# Matrix Multiplication (Row x Column dot product)
# [1*5 + 2*7, 1*6 + 2*8] = [19, 22]
# [3*5 + 4*7, 3*6 + 4*8] = [43, 50]
mat_prod = A @ B
print("\nMatrix Multiplication (A @ B):\n", mat_prod)
```

### 4. Transpose & Symmetric Matrix


```python
M = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("Original (2x3):\n", M)
print("\nTranspose (3x2):\n", M.T)

# Symmetric Matrix (where S == S.T)
S = np.array([
    [2, 3, 1],
    [3, 5, 4],
    [1, 4, 6]
])
is_symmetric = np.array_equal(S, S.T)
print("\nMatrix S:\n", S)
print("Is S symmetric (S == S.T)?:", is_symmetric)
```

### 5. Determinant of a Matrix


```python
# 2x2 Matrix: det = (ad - bc)
# [4, 2], [1, 3] => (4*3) - (2*1) = 12 - 2 = 10
A = np.array([[4, 2], [1, 3]])
det_A = np.linalg.det(A)

print("Matrix A:\n", A)
print("Determinant of A:", round(det_A, 2))

# Singular matrix example (det == 0, rows are linearly dependent)
Singular = np.array([[2, 4], [1, 2]])
det_Singular = np.linalg.det(Singular)
print("\nSingular Matrix:\n", Singular)
print("Determinant of Singular matrix:", round(det_Singular, 2))
```

### 6. Matrix Inverse


```python
A = np.array([[4, 2], [1, 3]])

# Inverse exists only if det != 0
A_inv = np.linalg.inv(A)
print("Matrix A:\n", A)
print("\nInverse (A^-1):\n", A_inv.round(3))

# Verification: A @ A_inv should be Identity matrix I
identity_check = A @ A_inv
print("\nVerification (A @ A_inv):\n", identity_check.round(1))
```

### 7. Trace & Diagonal Elements


```python
A = np.array([
    [5, 2, 1],
    [3, 8, 4],
    [6, 7, 9]
])

# Diagonal elements: [5, 8, 9]
diagonal = np.diag(A)
# Trace = sum of diagonal elements (5 + 8 + 9 = 22)
matrix_trace = np.trace(A)

print("Matrix A:\n", A)
print("Diagonal elements:", diagonal)
print("Trace (sum of diagonal):", matrix_trace)
```

### 8. Matrix Rank


```python
# Rank = number of linearly independent rows or columns
A = np.array([
    [1, 2, 3],
    [2, 4, 6],  # Row 2 is 2 * Row 1 (dependent)
    [5, 1, 0]
])

rank = np.linalg.matrix_rank(A)
print("Matrix A:\n", A)
print("Rank of A:", rank)  # Expected: 2 because row 2 is a multiple of row 1
```

### 9. Eigenvalues & Eigenvectors


```python
# For square matrix A: A @ v = lambda * v
A = np.array([
    [2, 0],
    [0, 5]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:\n", A)
print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)

# Verification for first eigenvalue & eigenvector: A @ v == lambda * v
v0 = eigenvectors[:, 0]
lam0 = eigenvalues[0]
print("\nA @ v0:", A @ v0)
print("lambda0 * v0:", lam0 * v0)
```

### 10. Solving Linear Systems of Equations (Ax = b)


```python
# System of Equations:
#  2x + 1y = 8
#  1x + 3y = 14

A = np.array([[2, 1], [1, 3]])
b = np.array([8, 14])

# Solve Ax = b
x = np.linalg.solve(A, b)

print("Coefficients Matrix A:\n", A)
print("Constants vector b:", b)
print(f"\nSolution: x = {x[0]:.1f}, y = {x[1]:.1f}")

# Check: A @ x == b
print("Verification (A @ x):", A @ x)
```

### 11. Singular Value Decomposition (SVD)


```python
# SVD decomposes matrix into: A = U @ diag(S) @ Vt
A = np.array([
    [3, 2, 2],
    [2, 3, -2]
])

U, s, Vt = np.linalg.svd(A)

print("Matrix A (2x3):\n", A)
print("\nLeft Singular Vectors U (2x2):\n", U.round(3))
print("\nSingular Values s:", s.round(3))
print("\nRight Singular Vectors Vt (3x3):\n", Vt.round(3))
```
