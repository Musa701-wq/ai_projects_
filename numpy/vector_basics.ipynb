# Vector Operations in NumPy


### 1. Vector Creation & Shapes


```python
import numpy as np

# 1D Vector (rank-1 array)
v = np.array([3, 4])
print("1D Vector:", v)
print("Shape:", v.shape)
print("Dimensions:", v.ndim)
print("Total elements:", v.size)

# Row vector (1, 3) and Column vector (3, 1)
row_vec = np.array([[1, 2, 3]])
col_vec = np.array([[1], [2], [3]])
print("\nRow vector shape:", row_vec.shape)
print("Column vector shape:", col_vec.shape)
```

### 2. Vector Addition & Subtraction


```python
u = np.array([2, 5])
v = np.array([3, 1])

print("Vector u:", u)
print("Vector v:", v)

# Element-wise addition & subtraction
add_res = u + v
sub_res = u - v

print("\nu + v:", add_res)
print("u - v:", sub_res)
```

### 3. Scalar Multiplication & Scaling


```python
v = np.array([2, 4])
scalar = 3

# Scaling magnitude
scaled = scalar * v

# Negative scalar reverses direction
reversed_v = -1 * v

print("Original vector:", v)
print("Scaled by 3:", scaled)
print("Reversed (-1 * v):", reversed_v)
```

### 4. Vector Magnitude (Norm / Length)


```python
v = np.array([3, 4])

# Euclidean norm (L2) formula: sqrt(x^2 + y^2)
manual_norm = np.sqrt(np.sum(v**2))

# Using numpy linalg norm (L2 Norm / Euclidean length)
l2_norm = np.linalg.norm(v)

# L1 Norm (Manhattan length: |x| + |y|)
l1_norm = np.linalg.norm(v, ord=1)

print("Vector:", v)
print("L2 Norm (manual):", manual_norm)
print("L2 Norm (np.linalg.norm):", l2_norm)
print("L1 Norm (Manhattan):", l1_norm)
```

### 5. Unit Vector (Normalization)


```python
v = np.array([3, 4])

# Unit vector = v / ||v|| (has magnitude of 1)
unit_v = v / np.linalg.norm(v)

print("Vector:", v)
print("Unit vector:", unit_v)
print("Magnitude of unit vector:", np.linalg.norm(unit_v))
```

### 6. Dot Product


```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

# Dot product: (1*4) + (2*5) + (3*6) = 4 + 10 + 18 = 32
dot1 = np.dot(u, v)
dot2 = u @ v

print("Vector u:", u)
print("Vector v:", v)
print("\nElement-wise product (u * v):", u * v)
print("Dot product (np.dot):", dot1)
print("Dot product (u @ v):", dot2)
```

### 7. Angle Between Vectors & Cosine Similarity


```python
u = np.array([1, 2])
v = np.array([2, 4])  # same direction, scaled

# Cosine similarity: (u . v) / (||u|| * ||v||)
cos_sim = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Angle in degrees
angle_rad = np.arccos(np.clip(cos_sim, -1.0, 1.0))
angle_deg = np.degrees(angle_rad)

print("Vector u:", u)
print("Vector v:", v)
print("Cosine Similarity:", round(cos_sim, 4))
print(f"Angle: {angle_deg:.2f} degrees")

# Perpendicular vectors example
a = np.array([1, 0])
b = np.array([0, 1])
cos_ab = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print("\nPerpendicular vectors cosine:", round(cos_ab, 4))
print(f"Angle: {np.degrees(np.arccos(cos_ab)):.2f} degrees")
```

### 8. Orthogonal Vectors (Perpendicularity)


```python
v1 = np.array([3, -2])
v2 = np.array([2, 3])

# If dot product is 0, vectors are orthogonal
dot_val = np.dot(v1, v2)
is_orthogonal = np.isclose(dot_val, 0)

print("v1:", v1)
print("v2:", v2)
print("Dot product:", dot_val)
print("Are vectors orthogonal?:", is_orthogonal)
```

### 9. Distance Between Vectors (Euclidean Distance)


```python
p1 = np.array([1, 2])
p2 = np.array([4, 6])

# Euclidean distance = ||p1 - p2||
dist_manual = np.sqrt(np.sum((p1 - p2)**2))
dist_norm = np.linalg.norm(p1 - p2)

print("Point 1:", p1)
print("Point 2:", p2)
print("Euclidean distance (manual):", dist_manual)
print("Euclidean distance (linalg.norm):", dist_norm)
```

### 10. Vector Projection


```python
# Projecting vector a onto vector b: (a . b / ||b||^2) * b
a = np.array([3, 4])
b = np.array([2, 0])

proj_a_on_b = (np.dot(a, b) / np.dot(b, b)) * b

print("Vector a:", a)
print("Vector b:", b)
print("Projection of a on b:", proj_a_on_b)
```

### 11. Cross Product (3D Vectors)


```python
# Cross product produces perpendicular vector in 3D
v1 = np.array([1, 0, 0])  # along X-axis
v2 = np.array([0, 1, 0])  # along Y-axis

cp = np.cross(v1, v2)

print("v1 (X-axis):", v1)
print("v2 (Y-axis):", v2)
print("Cross product (v1 x v2):", cp)
```
