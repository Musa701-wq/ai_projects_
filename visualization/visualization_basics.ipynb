# Data Visualization with Matplotlib & Seaborn


### 1. Setup & Styling


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Modern seaborn theme
sns.set_theme(style="whitegrid", palette="muted")
print("Matplotlib & Seaborn loaded successfully!")
```

### 2. Line Plots & Multi-line Subplots (Matplotlib)


```python
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun"])
sales_2025 = np.array([32, 45, 54, 60, 75, 88])
sales_2026 = np.array([40, 52, 63, 78, 92, 110])

plt.figure(figsize=(8, 4))
plt.plot(months, sales_2025, marker='o', linewidth=2, color='#1f77b4', label='2025 Sales')
plt.plot(months, sales_2026, marker='s', linewidth=2, linestyle='--', color='#ff7f0e', label='2026 Sales')

plt.title("Monthly Sales Growth Comparison", fontsize=13, fontweight='bold', pad=10)
plt.xlabel("Month", fontsize=11)
plt.ylabel("Sales ($k)", fontsize=11)
plt.legend(frameon=True)
plt.show()
```

### 3. Bar Charts & Grouped Comparisons


```python
courses = ["Python", "Data Science", "Machine Learning", "Deep Learning"]
students = [420, 310, 260, 180]

plt.figure(figsize=(7, 4))
bars = plt.bar(courses, students, color=['#4e79a7', '#f28e2b', '#e15759', '#76b7b2'], width=0.5)
plt.title("Course Enrollments 2026", fontsize=13, fontweight='bold', pad=10)
plt.xlabel("Course", fontsize=11)
plt.ylabel("Students Enrolled", fontsize=11)
plt.ylim(0, 500)

for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, y + 10, f"{y}", ha='center', va='bottom', fontsize=9, fontweight='bold')
plt.show()
```

### 4. Distribution Analysis: Histogram & KDE (Seaborn)


```python
np.random.seed(42)
scores = np.random.normal(loc=72, scale=10, size=300)

plt.figure(figsize=(7.5, 4))
sns.histplot(scores, kde=True, color='#2b5c8f', bins=15)
plt.axvline(np.mean(scores), color='red', linestyle='--', label=f"Mean: {np.mean(scores):.1f}")
plt.title("Exam Scores Normal Distribution & KDE", fontsize=13, fontweight='bold', pad=10)
plt.xlabel("Exam Score", fontsize=11)
plt.ylabel("Frequency", fontsize=11)
plt.legend()
plt.show()
```

### 5. Scatter Plot with Hue & Size (Seaborn)


```python
np.random.seed(42)
sample_data = pd.DataFrame({
    'Experience': np.random.randint(1, 12, 60),
    'Salary': np.random.randint(40000, 110000, 60),
    'Role': np.random.choice(['Developer', 'Data Analyst', 'ML Engineer'], 60),
    'Projects': np.random.randint(2, 15, 60)
})

plt.figure(figsize=(8, 4.5))
sns.scatterplot(
    data=sample_data,
    x='Experience',
    y='Salary',
    hue='Role',
    size='Projects',
    sizes=(40, 200),
    palette='Set1',
    alpha=0.85
)
plt.title("Experience vs Salary (by Role & Projects)", fontsize=13, fontweight='bold', pad=10)
plt.xlabel("Years of Experience", fontsize=11)
plt.ylabel("Salary ($)", fontsize=11)
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.show()
```

### 6. Box Plot & Outlier Detection (Seaborn)


```python
plt.figure(figsize=(7.5, 4))
sns.boxplot(data=sample_data, x='Role', y='Salary', palette='Pastel1')
plt.title("Salary Spread & Outliers by Role", fontsize=13, fontweight='bold', pad=10)
plt.xlabel("Role", fontsize=11)
plt.ylabel("Salary ($)", fontsize=11)
plt.show()
```

### 7. Correlation Heatmap (Seaborn)


```python
corr_matrix = sample_data[['Experience', 'Salary', 'Projects']].corr()

plt.figure(figsize=(5.5, 4))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True, linewidths=1)
plt.title("Correlation Matrix Heatmap", fontsize=13, fontweight='bold', pad=10)
plt.show()
```
